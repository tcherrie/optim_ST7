## 1) Function definition

We study :
- Ellipsoid : $f(x,y) \mapsto (x-1)^2 + 100(y-1)^2$
- Rosenbrock's banana : $f(x,y) \mapsto (x-1)^2 + 100(y-x^2)^2$

In both cases, the optimum point is $(x,y)^* = (1,1)$

```python
def f(x, rosenbrock = False):
    if rosenbrock:
        return (x[0]-1)**2 + 10 * (x[1]-x[0]**2)**2
    else:
        return (x[0]-1)**2 + 10 * (x[1]-1)**2
    

import numpy as np

def cplxStep(x, f, h = 1e-50):
    dfdx = np.zeros_like(x)
    for i in range(len(dfdx)):
        e_i = np.zeros_like(x)
        e_i[i] = 1
        dfdx[i] = f(x + 1j * e_i * h).imag / h
    return dfdx
```

```python
import matplotlib.pyplot as plt

def plot_contour(f, lb = 0, ub = 2):
    X = np.linspace(lb, ub, 100)
    Y = np.linspace(lb, ub, 100)
    X, Y = np.meshgrid(X, Y)
    Z = f([X,Y])
    _, ax = plt.subplots()
    CS = ax.contour(X, Y, Z, np.logspace(-2,3,20))
    plt.grid()
    ax.set_aspect('equal', 'box')

plot_contour(f)
```
