# Problem : optimization of an inductor

## 1) Context

We want to optimize an inductance like the one below. Optimization variables are the number of turns $N$ and the air gap $e$. The current $I = 1 \text{A}$ is fixed by the application (filtering), together with the cross-section of the magnetic core $S=1 \text{cm}^2$ (limited volume).




The inductance value $L$ is directly related to the size of the air gap and the number of turns. Ampere's theorem shows that it is equal to

$$ L(e, N) = \frac{\mu_0 S N^2}{e} $$

To avoid magnetic saturation, flux density must remain low $B_{max} \leq 0.5 \text{T}$, with

$$ B(e,N) = \frac{\mu_0 N I}{e}  $$


In addition to conventional DC copper losses, field leakage in conductors at the air gap leads to AC losses due to the proximity effect.




DC losses are proportional to the number of turns, while AC losses are proportional to the square of the induction field in the air gap. Total losses are limited by natural convection cooling. We should therefore have

$$ P(e,N) = R_0 N I^2 + \beta B(e,N)^2 \leq P_{max} = 10 W $$

with:
- $R_0 = 10^{-2} \Omega$
- $\beta = 10 W \cdot T^{-2}$ at low working frequency, or  $\beta = 100 W\cdot T^{-2}$ at higher working frequency.

_______
## 2) Optimization problem

We want to maximize the inductance, and respects saturation and losses constraints :

$$ 
\left\{
\begin{array}{ll}
\text{find}: & (e,N)^* = \arg \min_{e,N} f(e, N) = -L(e,N) \\
\text{subject to}: & 
\begin{array}{l}
B(e,N) - B_{max} \leq 0\\
P(e,N) - P_{max} \leq 0\\
\end{array}
\end{array}
\right.
$$
